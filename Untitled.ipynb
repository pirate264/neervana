```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("tips.csv")
```

```python
data.head(10)
```

```python
plt.figure(figsize=(6,4))
plt.scatter(data['day'], data['tip'])
plt.title("Scatter Plot of Tip vs Day")
plt.xlabel("Day")
plt.ylabel("Tip")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.plot(data['day'], data['tip'])
plt.title("Line Chart of Tip vs Day")
plt.xlabel("Day")
plt.ylabel("Tip")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.bar(data['day'], data['tip'])
plt.title("Bar Plot of Tip vs Day")
plt.xlabel("Day")
plt.ylabel("Tip")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.hist(data['total_bill'])
plt.title("Histogram of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

```python
product_df = pd.DataFrame({
    'Product': ['Laptop', 'Phone', 'Tablet', 'Headphones'],
    'Category': ['Electronics', 'Electronics', 'Electronics', 'Accessories'],
    'Price': [60000, 30000, 20000, 3000],
    'Stock': [10, 20, 15, 50]
})
```

```python
product_df['Discount'] = np.random.randint(5, 30, size=len(product_df))
```

```python
product_df['Final Price'] = product_df['Price'] - (product_df['Price'] * product_df['Discount'] / 100)
```

```python
product_df = product_df.drop(columns=['Stock'])
product_df
```

```python
student_df = pd.DataFrame({
    'Student': ['Amit', 'Riya', 'Karan', 'Neha', 'Rahul'],
    'Algorithm': [85, 78, 92, 45, 88],
    'Data Structure': [90, 82, 88, 60, 49],
    'English': [88, 91, 84, 55, 70]
})
```

```python
above_80 = student_df[
    (student_df['Algorithm'] > 80) &
    (student_df['Data Structure'] > 80) &
    (student_df['English'] > 80)
]
above_80
```

```python
below_50 = student_df[
    (student_df['Algorithm'] < 50) |
    (student_df['Data Structure'] < 50) |
    (student_df['English'] < 50)
]
below_50
```

```python
student_df['Total'] = student_df[['Algorithm', 'Data Structure', 'English']].sum(axis=1)

top_5_percent = student_df[
    student_df['Total'] >= student_df['Total'].quantile(0.95)
]
top_5_percent
```
