```python
import matplotlib.pyplot as plt

# Given dataset
x = [1, 2, 3, 4, 5]
y = [3, 7, 9, 13, 15]

n = len(x)

# Required summations
sum_x = sum(x)
sum_y = sum(y)
sum_x2 = sum(i**2 for i in x)
sum_xy = sum(x[i]*y[i] for i in range(n))

# Least Squares formulas
m = (n*sum_xy - sum_x*sum_y) / (n*sum_x2 - sum_x**2)
b = (sum_y - m*sum_x) / n

print("Slope (m):", m)
print("Intercept (b):", b)
```

```python
# Predicted values
y_pred = [m*i + b for i in x]

plt.scatter(x, y, color='blue', label="Actual Data")
plt.plot(x, y_pred, color='red', label="Regression Line")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Linear Regression using Least Squares")
plt.legend()
plt.show()

```

```python
print("x  Actual y  Predicted y")
for i in range(n):
    print(x[i], " ", y[i], " ", round(y_pred[i], 2))
```

```python
sse = sum((y[i] - y_pred[i])**2 for i in range(n))
print("Sum of Squared Errors:", sse)
```

```python
plt.plot(x, y_pred, color='green', label="f_new(x)")
plt.scatter(x, y, label="Actual Data")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
x_new = 4.5
y_new = m*x_new + b
print("Predicted value for x = 4.5:", y_new)

plt.scatter(x, y, label="Actual Data")
plt.plot(x, y_pred, label="Regression Line")
plt.scatter(x_new, y_new, color='purple', label="Predicted Point")
plt.legend()
plt.show()
```

```python
import numpy as np
import pandas as pd

# Generate random data
np.random.seed(0)
x = np.random.uniform(40, 50, 100)
y = 3*x + 0.4 + np.random.normal(0, 2, 100)

# Save to CSV
df = pd.DataFrame({'x': x, 'y': y})
df.to_csv("random_data.csv", index=False)

# Read CSV
data = pd.read_csv("random_data.csv")

x = data['x']
y = data['y']
n = len(x)
```

```python
m = (n*(x*y).sum() - x.sum()*y.sum()) / (n*(x**2).sum() - (x.sum())**2)
b = (y.sum() - m*x.sum()) / n

print("Slope (m):", m)
print("Intercept (b):", b)

y_pred = m*x + b
```

```python
plt.scatter(x, y, label="CSV Data")
plt.plot(x, y_pred, color='red', label="Regression Line")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
x_unknown = 43.5
y_unknown = m*x_unknown + b
print("Predicted value for x = 43.5:", y_unknown)

plt.scatter(x, y, label="CSV Data")
plt.plot(x, y_pred, label="Regression Line")
plt.scatter(x_unknown, y_unknown, color='green', label="Predicted Point")
plt.legend()
plt.show()
```
