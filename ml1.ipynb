```python
import pandas as pd

data = {
    "Name": ["Aman", "Riya", "Karan", "Sneha", "Rahul"],
    "Score": [72, 55, 81, 60, 45]
}

df = pd.DataFrame(data)
print("Original DataFrame:\n", df)

filtered_df = df[df["Score"] >= 60]
print("\nFiltered DataFrame (Score >= 60):\n", filtered_df)
```

```python
def normalize(data):
    min_val = min(data)
    max_val = max(data)
    return [(x - min_val) / (max_val - min_val) for x in data]

# Example usage
numbers = [10, 20, 30, 40, 50]
normalized_numbers = normalize(numbers)
print("Normalized values:", normalized_numbers)
```
